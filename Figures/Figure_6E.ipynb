## Selectivity 0.6

```python
bold_data_successful = np.load('saved_plots/Figure_6D/bold_signal_0.6.npy', allow_pickle=True) # Data for successful trial
fr_data_successful = np.load('saved_plots/Figure_6D/activity_trace_fr_0.6.npy', allow_pickle=True) # Data for successful trial
dt = 0.5e-3 # Time resolution of simulation
bold_data_successful.shape
bold_A_data_successful = np.array(bold_data_successful[:, 0])
bold_A_data_successful.shape
```

```python
# Data when stimulus was given to population A
time_to_cut = 10 # s
index_area_to_decode = 8
samples_to_cut = int(time_to_cut/dt)
bold_A_data_successful = np.array(bold_data_successful[:, 0])
bold_A_data_successful = np.array([trial[samples_to_cut:, index_area_to_decode] for trial in bold_A_data_successful])
bold_B_data_successful = np.array(bold_data_successful[:, 1])
bold_B_data_successful = np.array([trial[samples_to_cut:, index_area_to_decode] for trial in bold_B_data_successful])

standard_deviation_bold_A_successful = np.std(bold_A_data_successful, axis=0, ddof=1)
mean_bold_A_successful = np.mean(bold_A_data_successful, axis=0)
standard_deviation_bold_B_successful = np.std(bold_B_data_successful, axis=0, ddof=1)
mean_bold_B_successful = np.mean(bold_B_data_successful, axis=0)

fr_A_data_successful = np.array(fr_data_successful[:, 0])
fr_A_data_successful = np.array([trial[samples_to_cut:, index_area_to_decode] for trial in fr_A_data_successful])
fr_B_data_successful = np.array(fr_data_successful[:, 1])
fr_B_data_successful = np.array([trial[samples_to_cut:, index_area_to_decode] for trial in fr_B_data_successful])
standard_deviation_fr_A_successful = np.std(fr_A_data_successful, axis=0, ddof=1)
mean_fr_A_successful = np.mean(fr_A_data_successful, axis=0)
standard_deviation_fr_successful = np.std(fr_B_data_successful, axis=0, ddof=1)
mean_fr_B_successful = np.mean(fr_B_data_successful, axis=0)
```

```python
import matplotlib as mpl
mpl.rcParams.update({
"font.size": 10,
"axes.labelsize": 9,
"axes.titlesize": 10,
"xtick.labelsize": 9,
"ytick.labelsize": 9,
"legend.fontsize": 9,
"axes.linewidth": 0.8,
"xtick.major.width": 0.8,
"ytick.major.width": 0.8,
"xtick.major.size": 3,
"ytick.major.size": 3,
})
fig, ax = plt.subplots(figsize=(2.2, 1.9))
ax.plot(np.arange(0, len(mean_bold_A_successful))*dt, mean_bold_A_successful, label='Population A', color='#56c68a');
ax.plot(np.arange(0, len(mean_bold_B_successful))*dt, mean_bold_B_successful, label='Population B', color='#eaa8b7');
ax.fill_between(np.arange(0, len(mean_bold_A_successful))*dt, mean_bold_A_successful - standard_deviation_bold_A_successful,
                mean_bold_A_successful + standard_deviation_bold_A_successful, alpha=0.2, color='#56c68a')
ax.fill_between(np.arange(0, len(mean_bold_B_successful))*dt, mean_bold_B_successful - standard_deviation_bold_B_successful,
                mean_bold_B_successful + standard_deviation_bold_B_successful, alpha=0.2, color='#eaa8b7')
ax.set_xlabel('Time (s)')
ax.set_ylabel('BOLD Signal')
ax.spines[['right', 'top']].set_visible(False)
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.grid(linestyle=':', alpha=0.5)
ax.tick_params(axis='both', which='major')
ax.tick_params(axis='both', which='minor')
ax.set_title('')
plt.savefig('saved_plots/Figure_6E/bold_decoding_trace_selectivity_06.pdf')
```

```python
fig, ax = plt.subplots(figsize=(2.2, 1.9))
ax.plot(np.arange(0, len(mean_fr_A_successful))*dt, mean_fr_A_successful, label='Population A', color='#56c68a');
ax.plot(np.arange(0, len(mean_fr_B_successful))*dt, mean_fr_B_successful, label='Population B', color='#eaa8b7');
ax.set_ylim(-2, 60)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Firing Rate (Hz)')
ax.spines[['right', 'top']].set_visible(False)
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.grid(linestyle=':', alpha=0.5)
ax.tick_params(axis='both', which='major')
ax.tick_params(axis='both', which='minor')
ax.set_title('')
plt.savefig('saved_plots/Figure_6E/fr_decoding_trace_selectivity_06.pdf')
```

## Four standalone panels: BOLD and firing rate traces (journal format)

Each panel is a fourth of the previous 4.6 x 4.6 cm combined figure: 2.3 x 2.3 cm.
Saved as four separate svg files for external assembly. Loads its own data from
saved_plots/Figure_6D, so it does not depend on the cells above.

```python
# Four standalone panels (BOLD x2, firing rate x2), each a fourth of the previous
# combined-panel space: 0.25 x 18.4 / 2 = 2.3 cm wide by 4.6 / 2 = 2.3 cm tall.
# Each is saved as its own svg, to be assembled together outside the notebook.
#
# Font is 6 pt (5.5 pt for the title, which is the one label at risk of overflowing at
# this width) with 2 tick marks per axis - there isn't room for more at 2.3 x 2.3 cm.
# No legend, as in the combined panel above: Population A is green (#56c68a), B is pink
# (#eaa8b7), consistent with the other trace panels in this project.
#
# Data is reloaded from the saved .npy files (rather than reusing the notebook's
# mean_bold_A_successful etc.), because those names get overwritten when the
# "Selectivity 1" section runs after "Selectivity 0.6".
import matplotlib as mpl
from matplotlib.ticker import MaxNLocator


def load_selectivity_traces(select):
    """Same processing as the 'Selectivity <x>' cells above: load, cut, and average."""
    bold_data = np.load(f'saved_plots/Figure_6D/bold_signal_{select}.npy', allow_pickle=True)
    fr_data = np.load(f'saved_plots/Figure_6D/activity_trace_fr_{select}.npy', allow_pickle=True)
    dt = 0.5e-3
    time_to_cut = 10
    index_area_to_decode = 8
    samples_to_cut = int(time_to_cut / dt)

    bold_A = np.array([trial[samples_to_cut:, index_area_to_decode] for trial in np.array(bold_data[:, 0])])
    bold_B = np.array([trial[samples_to_cut:, index_area_to_decode] for trial in np.array(bold_data[:, 1])])
    fr_A = np.array([trial[samples_to_cut:, index_area_to_decode] for trial in np.array(fr_data[:, 0])])
    fr_B = np.array([trial[samples_to_cut:, index_area_to_decode] for trial in np.array(fr_data[:, 1])])

    return dict(
        dt=dt,
        mean_bold_A=bold_A.mean(0), std_bold_A=bold_A.std(0, ddof=1),
        mean_bold_B=bold_B.mean(0), std_bold_B=bold_B.std(0, ddof=1),
        mean_fr_A=fr_A.mean(0), mean_fr_B=fr_B.mean(0),
    )


traces_by_selectivity = [('0.6', load_selectivity_traces('0.6')), ('1', load_selectivity_traces('1'))]

fig_params = {
    "text.usetex": False,
    "font.family": "sans-serif",
    "font.sans-serif": ["Open Sans"],
    "font.size": 6,
    "axes.labelsize": 6,
    "axes.titlesize": 5.5,
    "xtick.labelsize": 6,
    "ytick.labelsize": 6,
    "axes.linewidth": 0.7,
    "xtick.major.width": 0.7,
    "ytick.major.width": 0.7,
    "xtick.major.size": 1.5,
    "ytick.major.size": 1.5,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "svg.fonttype": "none",
}

CM = 2.54
W = (0.25 * 18.4) / 2 / CM   # a fourth of the previous 4.6 x 4.6 cm panel
H = 4.6 / 2 / CM


def style_axes(ax):
    ax.minorticks_off()
    ax.xaxis.set_major_locator(MaxNLocator(2))
    ax.yaxis.set_major_locator(MaxNLocator(2))
    ax.grid(linestyle=':', alpha=0.5)
    for spine in ['top', 'right']:
        ax.spines[spine].set_visible(False)
    ax.xaxis.set_ticks_position('bottom')
    ax.yaxis.set_ticks_position('left')
    # y-tick numbers dropped: "0.008" (BOLD) and "6" (FR) reserve different label
    # widths, which shrinks the BOLD plot box relative to the FR one at fixed figure
    # size. Ticks stay so the scale is still visible; values are placed by hand later.
    # ax.tick_params(axis='y', labelleft=False)


for sel, d in traces_by_selectivity:
    t = np.arange(len(d['mean_bold_A'])) * d['dt']

    # --- BOLD panel ---
    with mpl.rc_context(fig_params):
        fig, ax = plt.subplots(figsize=(W, H), constrained_layout=True)
        ax.plot(t, d['mean_bold_A'], color='#56c68a', lw=0.7)
        ax.plot(t, d['mean_bold_B'], color='#eaa8b7', lw=0.7)
        ax.fill_between(t, d['mean_bold_A'] - d['std_bold_A'], d['mean_bold_A'] + d['std_bold_A'],
                        alpha=0.2, color='#56c68a')
        ax.fill_between(t, d['mean_bold_B'] - d['std_bold_B'], d['mean_bold_B'] + d['std_bold_B'],
                        alpha=0.2, color='#eaa8b7')
        # ax.set_title(f'Selectivity {sel}')
        # ax.set_xlabel('Time (s)')
        # ax.set_ylabel('BOLD Signal')
        style_axes(ax)

        # no bbox_inches='tight', so the file keeps the exact 2.3 x 2.3 cm size
        fig.savefig(f'saved_plots/Figure_6E/bold_trace_selectivity_{sel}_panel.pdf')
        plt.show()

    # --- Firing rate panel ---
    with mpl.rc_context(fig_params):
        fig, ax = plt.subplots(figsize=(W, H), constrained_layout=True)
        ax.plot(t, d['mean_fr_A'], color='#56c68a', lw=0.7)
        ax.plot(t, d['mean_fr_B'], color='#eaa8b7', lw=0.7)
        ax.set_ylim(-2, 60)
        # ax.set_xlabel('Time (s)')
        # ax.set_ylabel('Firing Rate (Hz)')
        style_axes(ax)

        fig.savefig(f'saved_plots/Figure_6E/fr_trace_selectivity_{sel}_panel.svg')
        plt.show()
```
